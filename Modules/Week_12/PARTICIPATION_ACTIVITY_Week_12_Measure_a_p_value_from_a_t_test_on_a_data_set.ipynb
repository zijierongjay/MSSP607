```python
import pandas as pd

students = pd.read_csv(
    'Modules/Week_12/Student Performance Data.csv'
)

students.head()
```

```python
students.shape
students.columns
students.info()
students.isnull().sum()
```

```python
students['gender'].value_counts()
```

```python
students['test preparation course'].value_counts()
```

```python
students[
    students['gender'] == 'male'
]['test preparation course'].value_counts()
```

```python
male_completed = students[
    (students['gender'] == 'male') &
    (students['test preparation course'] == 'completed')
]['total score']

male_none = students[
    (students['gender'] == 'male') &
    (students['test preparation course'] == 'none')
]['total score']
```

```python
print("Male completed mean:", male_completed.mean())
print("Male none mean:", male_none.mean())
```

```python
print("Male completed n:", len(male_completed))
print("Male none n:", len(male_none))
```

```python
from scipy import stats

print(stats.normaltest(male_completed))
print(stats.normaltest(male_none))
```

```python
t_stat, p_value = stats.ttest_ind(
    male_completed,
    male_none,
    equal_var=False
)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

2

Yes, there is a statistically significant difference between male students who completed the test preparation course and those who did not. The average total score for males who completed the course was about 212.34, while the average for males who did not complete it was about 189.13. The p-value from the t-test was much smaller than 0.05, so the difference is statistically significant. This suggests that the male students who completed the preparation course generally had higher total scores.

```python
students['lunch'].value_counts()
```

```python
standard_lunch = students[
    students['lunch'] == 'standard'
]['total score']

free_reduced_lunch = students[
    students['lunch'] == 'free/reduced'
]['total score']
```

```python
print("Standard lunch mean:", standard_lunch.mean())
print("Free/reduced lunch mean:", free_reduced_lunch.mean())
```

```python
t_stat, p_value = stats.ttest_ind(
    standard_lunch,
    free_reduced_lunch,
    equal_var=False
)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

Another difference I found was between students with standard lunch and students with free/reduced lunch. Students with standard lunch had an average total score of about 212.51, while students with free/reduced lunch had an average score of about 186.60. The p-value from the t-test was much smaller than 0.05, so the difference is statistically significant. This suggests that students with standard lunch generally had higher total scores in this dataset.
