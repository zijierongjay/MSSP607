2a

```python
japan_mb = df[
    (df['parsed_place'].str.endswith('Japan')) &
    (df['magType'] == 'mb')
]

japan_mb['mag'].quantile(0.95)
```

The 95th percentile of earthquake magnitude in Japan for mb earthquakes is 4.9. This means 95% of these earthquakes had magnitudes at or below 4.9.

2b

```python
indonesia = df[
    df['parsed_place'].str.endswith('Indonesia')
]

indonesia['tsunami'].value_counts(normalize=True)
```

```python
tsunami_percentage = indonesia['tsunami'].mean()

print(f"Percentage of earthquakes in Indonesia with tsunamis: {tsunami_percentage:.2%}")
```

About 23.13% of earthquakes in Indonesia were associated with tsunamis. This means roughly one out of every four earthquakes in the dataset had a tsunami.

```python
df['ring_of_fire'] = df.parsed_place.str.contains(r'|'.join([
    'Bolivia', 'Chile', 'Ecuador', 'Peru', 'Costa Rica',
    'Guatemala', '^Mexico', 'Japan', 'Philippines', 'Indonesia',
    'New Zealand', 'Antarctic', 'Canada', 'Fiji',
    'Alaska', 'Washington', 'California', 'Russia',
    'Taiwan', 'Tonga', 'Kermadec Islands'
]))

df[['ring_of_fire', 'parsed_place']]
```

2e

```python
earthquakes = df[df['type'] == 'earthquake']

earthquakes['ring_of_fire'].value_counts()
```

There were 7,004 earthquakes in Ring of Fire locations and 2,077 earthquakes outside the Ring of Fire.

2f

```python
tsunami_count = df.loc[df['ring_of_fire'], 'tsunami'].sum()

print("Tsunami count along the Ring of Fire:", tsunami_count)
```

There were 45 tsunamis along the Ring of Fire.
