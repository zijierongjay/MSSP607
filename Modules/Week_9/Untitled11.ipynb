```python
import requests
import bs4
import re
from bs4 import BeautifulSoup

url = 'https://www.flickr.com/commons/'

response = requests.get(url)

print(response.status_code)
print(response.status_code == requests.codes.ok)
```

```python
soup = bs4.BeautifulSoup(
    response.content,
    'html.parser'
)

print(type(soup))
```

```python
images = soup.find_all('img')

print("Number of image tags:", len(images))
```

```python
for img in images:
    print(img.get('src'))
```

```python
image_urls = []

for img in images:
    src = img.get('src')

    if src and ('.jpg' in src or '.jpeg' in src):
        if src.startswith('//'):
            src = 'https:' + src

        image_urls.append(src)

image_urls
```

```python
img_response = requests.get(image_urls[0])

print(img_response.status_code)
```

```python
with open('flickr_image_1.jpg', 'wb') as file:
    file.write(img_response.content)

print('Image downloaded successfully.')
```

```python
search_term = 'Philadelphia'
```

```python
search_url = 'https://www.flickr.com/search/?text=' + search_term

print(search_url)
```

```python
search_response = requests.get(search_url)

print(search_response.status_code)
```

```python
search_soup = BeautifulSoup(
    search_response.content,
    'html.parser'
)

search_images = search_soup.find_all('img')

print("Number of image tags:", len(search_images))
```

```python
for img in search_images:
    print(img.get('src'))
```

```python
philadelphia_urls = []

for img in search_images:
    src = img.get('src')

    if src and ('.jpg' in src or '.jpeg' in src):
        if src.startswith('//'):
            src = 'https:' + src

        philadelphia_urls.append(src)

philadelphia_urls
```

```python
print("Number of Philadelphia images:", len(philadelphia_urls))
```

```python
import os

folder_name = 'Philadelphia_Images'

os.makedirs(folder_name, exist_ok=True)
```

```python
for i, url in enumerate(philadelphia_urls, start=1):

    response = requests.get(url)

    if response.status_code == 200:

        file_name = f'{folder_name}/philadelphia_{i}.jpg'

        with open(file_name, 'wb') as file:
            file.write(response.content)

        print(f'Downloaded: {file_name}')
```

```python
import shutil

shutil.make_archive(
    'Philadelphia_Images',
    'zip',
    'Philadelphia_Images'
)
```
