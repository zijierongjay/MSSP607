```python
import pandas as pd

aapl = pd.read_csv('Repository data/Week 05 Examples/aapl.csv')
amzn = pd.read_csv('Repository data/Week 05 Examples/amzn.csv')
nflx = pd.read_csv('Repository data/Week 05 Examples/nflx.csv')
fb = pd.read_csv('Repository data/Week 05 Examples/fb.csv')
goog = pd.read_csv('Repository data/Week 05 Examples/goog.csv')
```

2c

```python
aapl['ticker'] = 'AAPL'
amzn['ticker'] = 'AMZN'
nflx['ticker'] = 'NFLX'
fb['ticker'] = 'FB'
goog['ticker'] = 'GOOG'
```

```python
aapl.head()
```

```python
print(aapl.ticker.unique())
print(amzn.ticker.unique())
print(nflx.ticker.unique())
print(fb.ticker.unique())
print(goog.ticker.unique())
```

```python
faang = pd.concat([
    fb,
    aapl,
    amzn,
    nflx,
    goog
])

faang.head()
```

```python
faang.ticker.value_counts()
```

```python
faang.to_csv('faang.csv', index=False)

print("faang.csv saved.")
```

2f

```python
faang = pd.read_csv('faang.csv')

faang.head()
```

```python
faang.dtypes
```

```python
faang['date'] = pd.to_datetime(faang['date'])
faang['volume'] = faang['volume'].astype(int)

faang.dtypes
```

```python
faang = faang.sort_values(
    by=['date', 'ticker']
)

faang.head(10)
```

2g

```python
highest_volume = faang.nlargest(7, 'volume')

highest_volume
```

2h

```python
faang_melted = faang.melt(
    id_vars=['date', 'ticker']
)

faang_melted.head(10)
```
