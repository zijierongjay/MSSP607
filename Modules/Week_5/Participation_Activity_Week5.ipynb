```python
import re

notebook_path = '/content/MSSP607/Modules/Week_5/Participation_Activity_Week5.ipynb'

with open(notebook_path, 'r', encoding='utf-8') as file:
    notebook_text = file.read()

token_pattern = r'github_pat_[A-Za-z0-9_]{20,}|ghp_[A-Za-z0-9]{20,}'

tokens_found = re.findall(token_pattern, notebook_text)

print("Number of tokens found:", len(tokens_found))

cleaned_text = re.sub(
    token_pattern,
    '[REDACTED]',
    notebook_text
)

with open(notebook_path, 'w', encoding='utf-8') as file:
    file.write(cleaned_text)

print("Notebook cleaned.")
```

```python
with open(notebook_path, 'r', encoding='utf-8') as file:
    notebook_text = file.read()

token_found = bool(
    re.search(
        r'github_pat_[A-Za-z0-9_]{20,}|ghp_[A-Za-z0-9]{20,}',
        notebook_text
    )
)

print("Token still found:", token_found)
```
