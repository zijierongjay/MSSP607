## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(list(salaries))
```

## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
mean = sum(salaries) / len(salaries)
print(mean)
```

```python
import statistics

print(statistics.mean(salaries))
```

### median

```python
sorted_salaries = sorted(salaries)
n = len(sorted_salaries)

if n % 2 == 0:
    median = (sorted_salaries[n//2 - 1] + sorted_salaries[n//2]) / 2
else:
    median = sorted_salaries[n//2]

print(median)
```

```python
import statistics
print(statistics.median(salaries))
```

### mode

```python
mode = max(salaries, key=salaries.count)
print(mode)
```

```python
print(statistics.mode(salaries))
```

### sample variance
Remember to use Bessel's correction.

```python
import statistics
mean = sum(salaries) / len(salaries)
sample_variance = sum((x - mean) ** 2 for x in salaries) / (len(salaries) - 1)
print("Sample variance:", sample_variance)
```

```python
print("Verified variance:", statistics.variance(salaries))
```

### sample standard deviation
Remember to use Bessel's correction.

```python
sample_std = sample_variance ** 0.5
print("Sample standard deviation:", sample_std)
```

```python
print("Verified standard deviation:", statistics.stdev(salaries))
```

## Exercise 6: Calculating more statistics
### range

```python
import numpy as np
salary_range = max(salaries) - min(salaries)
print("Range:", salary_range)
```

```python
print("Verified range:", np.ptp(salaries))
```

### coefficient of variation

```python
import statistics
mean = sum(salaries) / len(salaries)
sample_variance = sum((x - mean) ** 2 for x in salaries) / (len(salaries) - 1)
sample_std = sample_variance ** 0.5
cv = sample_std / mean
print("Coefficient of variation:", cv)
```

```python
verified_cv = statistics.stdev(salaries) / statistics.mean(salaries)
print("Verified coefficient of variation:", verified_cv)
```

### interquartile range

```python
import numpy as np
from scipy.stats import iqr
q1 = np.percentile(salaries, 25)
q3 = np.percentile(salaries, 75)
interquartile_range = q3 - q1
print("Q1:", q1)
print("Q3:", q3)
print("Interquartile range:", interquartile_range)
```

```python
print("Verified IQR:", iqr(salaries))
```

### quartile coefficent of dispersion

```python
qcd = (q3 - q1) / (q3 + q1)
print("Quartile coefficient of dispersion:", qcd)
```

```python
verified_q1 = np.percentile(salaries, 25)
verified_q3 = np.percentile(salaries, 75)
verified_qcd = (verified_q3 - verified_q1) / (verified_q3 + verified_q1)
print("Verified quartile coefficient of dispersion:", verified_qcd)
```

## Exercise 7: Scaling data
### min-max scaling

```python
minimum = min(salaries)
maximum = max(salaries)

min_max_scaled = [
    (x - minimum) / (maximum - minimum)
    for x in salaries
]

print(min_max_scaled)
```

```python
import numpy as np

verified = (
    (np.array(salaries) - np.min(salaries))
    / (np.max(salaries) - np.min(salaries))
)

print(verified)
```

### standardizing

```python
import statistics

mean = sum(salaries) / len(salaries)

sample_variance = sum(
    (x - mean) ** 2 for x in salaries
) / (len(salaries) - 1)

sample_std = sample_variance ** 0.5

standardized = [
    (x - mean) / sample_std
    for x in salaries
]

print(standardized)
```

```python
from scipy.stats import zscore

verified = zscore(salaries, ddof=1)

print(verified)
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
import statistics

mean_x = sum(salaries) / len(salaries)
mean_y = sum(standardized) / len(standardized)

covariance = sum(
    (x - mean_x) * (y - mean_y)
    for x, y in zip(salaries, standardized)
) / (len(salaries) - 1)

print("Covariance:", covariance)
```

```python
print("Verified covariance:",
      statistics.covariance(salaries, standardized))
```

### Pearson correlation coefficient ($\rho$)

```python
std_x = statistics.stdev(salaries)
std_y = statistics.stdev(standardized)

correlation = covariance / (std_x * std_y)

print("Pearson correlation:", correlation)
```

```python
print("Verified correlation:",
      statistics.correlation(salaries, standardized))
```
