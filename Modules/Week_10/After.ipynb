```python
import random

random.seed(0)

number_of_salaries = 100
max_salary = 1_000_000

salaries = [
    round(random.uniform(0, max_salary), 3)
    for _ in range(number_of_salaries)
]

print(salaries)
```

```python
import statistics

average_salary = statistics.mean(salaries)

print(f"Average salary: {average_salary:,.2f}")
```

### median

```python
import statistics

median_salary = statistics.median(salaries)

print(f"Median salary: {median_salary:,.2f}")
```

### mode

```python
import statistics

mode_salary = statistics.mode(salaries)

print(f"Mode salary: {mode_salary:,.2f}")
```

### sample variance
Remember to use Bessel's correction.

```python
import statistics

sample_variance = statistics.variance(salaries)

print(f"Sample variance: {sample_variance:,.2f}")
```

### sample standard deviation
Remember to use Bessel's correction.

```python
sample_std = sample_variance ** 0.5
print("Sample standard deviation:", sample_std)
```

```python
print("Verified standard deviation:", statistics.stdev(salaries))
```

## Exercise 6: Calculating more statistics
### range

```python
import statistics

sample_std = statistics.stdev(salaries)

print(f"Sample standard deviation: {sample_std:,.2f}")
```

### coefficient of variation

```python
import statistics

average_salary = statistics.mean(salaries)
sample_std = statistics.stdev(salaries)

cv = sample_std / average_salary

print(f"Coefficient of variation: {cv:.4f}")
```

### interquartile range

```python
import numpy as np
from scipy.stats import iqr

q1, q3 = np.percentile(salaries, [25, 75])
interquartile_range = iqr(salaries)

print(f"Q1: {q1:,.2f}")
print(f"Q3: {q3:,.2f}")
print(f"Interquartile range: {interquartile_range:,.2f}")
```

### quartile coefficent of dispersion

```python
qcd_percent = ((q3 - q1) / (q3 + q1)) * 100

print(f"Quartile coefficient of dispersion: {qcd_percent:.2f}%")
```

## Exercise 7: Scaling data
### min-max scaling

```python
minimum = min(salaries)
maximum = max(salaries)

min_max_scaled = [
    (x - minimum) / (maximum - minimum)
    for x in salaries
]

print(min_max_scaled)
```

```python
import numpy as np

verified = (
    (np.array(salaries) - np.min(salaries))
    / (np.max(salaries) - np.min(salaries))
)

print(verified)
```

### standardizing

```python
import statistics

mean = sum(salaries) / len(salaries)

sample_variance = sum(
    (x - mean) ** 2 for x in salaries
) / (len(salaries) - 1)

sample_std = sample_variance ** 0.5

standardized = [
    (x - mean) / sample_std
    for x in salaries
]

print(standardized)
```

```python
from scipy.stats import zscore

verified = zscore(salaries, ddof=1)

print(verified)
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
import statistics

mean_x = sum(salaries) / len(salaries)
mean_y = sum(standardized) / len(standardized)

covariance = sum(
    (x - mean_x) * (y - mean_y)
    for x, y in zip(salaries, standardized)
) / (len(salaries) - 1)

print("Covariance:", covariance)
```

```python
print("Verified covariance:",
      statistics.covariance(salaries, standardized))
```

### Pearson correlation coefficient ($\rho$)

```python
std_x = statistics.stdev(salaries)
std_y = statistics.stdev(standardized)

correlation = covariance / (std_x * std_y)

print("Pearson correlation:", correlation)
```

```python
print("Verified correlation:",
      statistics.correlation(salaries, standardized))
```
