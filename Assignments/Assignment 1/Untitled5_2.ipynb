```python
mother_score = df.groupby('EDU_MOTHER').G_SC.mean()

mother_score.sort_values(ascending=False)
```

```python
highest_mother_score = mother_score_clean.max()
lowest_mother_score = mother_score_clean.min()

difference_mother = highest_mother_score - lowest_mother_score

print("Highest average score:", highest_mother_score)
print("Lowest average score:", lowest_mother_score)
print("Difference:", difference_mother)
```

Parental education appears to be associated with student performance.
